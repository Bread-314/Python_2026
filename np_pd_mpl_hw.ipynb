## 1) NumPy
В эксперименте измерили уровень экспрессии одного гена в пяти образцах. Необходимо проанализировать результаты с помощью NumPy. 

Что нужно сделать:
- Вычислить среднее значение экспрессии
- Найти минимальное и максимальное значения
- Определить индекс образца с максимальной экспрессией
- Создать новый массив, содержащий только значения выше среднего
- Отсортировать исходные значения по возрастанию и вывести результат

```python
import numpy as np
expression = np.array([12, 18, 9, 25, 16])
mean_val = np.mean(expression)
min_val, max_val = np.min(expression), np.max(expression)
max_index_expr = np.argmax (expression)
higher_expr = expression[expression > mean_val]
asc_expr = np.sort(expression)

mean_val, higher_expr, asc_expr
```

## 2) Pandas
Исследователь измерил экспрессию гена в образцах двух экспериментальных групп. Нужно очистить данные и сравнить группы. 

Что нужно сделать:
1) Вывести первые три строки таблицы.
2) Найти пропущенные значения в таблице.
3) Удалить строки, в которых отсутствует значение экспрессии.
4) Отобрать образцы с экспрессией не ниже 18.
5) Рассчитать среднюю экспрессию отдельно для групп A и B.
Ожидаемый результат: очищенная таблица, список образцов с высокой экспрессией и средние значения для каждой группы.
Подсказка: используйте head(), isna(), dropna(), фильтрацию по условию и groupby() с mean()

```python
import pandas as pd
df = pd.DataFrame({
    "Образец": ["A1", "A2", "A3", "B1", "B2", "B3"],
    "Группа": ["A", "A", "A", "B", "B", "B"],
    "Экспрессия": [10, 15, None, 20, 25, 18]
})
df.head(3)
df[df.isna().any(axis = 1)]
df = df.dropna()
df[df['Экспрессия'] >= 18]
df.groupby('Группа')['Экспрессия'].mean()
```

Постройте график изменения экспрессии гена в зависимости от времени наблюдения. 

Что нужно сделать:
- Построить линейный график зависимости экспрессии от времени.
- Подписать оси: «Время, ч» и «Экспрессия гена».
- Добавить заголовок «Изменение экспрессии гена».
- Отобразить точки измерений маркерами.
- Включить сетку и показать график.
Подсказка: используйте plt.plot(), plt.xlabel(), plt.ylabel(), plt.title(), plt.grid() и plt.show().

```python
import matplotlib.pyplot as plt
time = [0, 2, 4, 6, 8, 10]
expression = [5, 8, 12, 11, 17, 23]
plt.plot(time, expression, marker = 'o')
plt.xlabel('Время, ч'), plt.ylabel('Экспрессия гена'), plt.title('Изменение экспрессии гена')
plt.grid()
plt.show()
```
