## While Loop
The while loop is one of the first loops that we'll probably encounter when we're starting to learn how to program.

There are three components that we need to construct the while loop in Python:

> The 'while' keyword

> A condition that transates to either 'True' or 'False'

> A block of code that we want to execute repeatedly.

```python
print("Thank you")
print("Thank you")
```

```python
print("Thank you")
print("Thank you")
print("Thank you")
print("Thank you")
print("Thank you")
print("Thank you")
print("Thank you")
print("Thank you")
print("Thank you")
print("Thank you")
```

```python
# syntax of while loop
number = 1
while number <= 10:
    print("Thank you")

    # increment the value of the variable "number by 1"
    number += 1  # number = number + 1

print("DONE!")
```

```python
# syntax of if statement
number = 1
if number <= 10:
    print("Thank you")

    # increment the value of the variable "number by 1"
    number += 1  # number = number + 1

print("DONE!")
```

```python
# Explanation of while loop:
# number = 2

# while number < 5:  # 2 < 5 => True
# print("Thank you")
# number = 3

# while number < 5:  # 3 < 5 => True
# print("Thank you")
# number = 4

# while number < 5:  # 4 < 5 => True
# print("Thank you")
# number = 5

# while number < 5:  # 5 < 5 => False
# Done!
```

```python
number = 1
while number <= 10:
    if number % 2 == 0:
        print("The number", number, "is even")
    else:
        print("The number", number, "is odd")

    number += 1
```

```python
# Explanation of the example
# # iteration 1
# number = 2
# while number < 5:    # 2 < 5 => True
# if number % 2 == 0:  # 2 % 2 == 0 => True
# print("The number", number, "is even")
# number = 3

# # iteration 2
# while number < 5:    # 3 < 5 => True
# if number % 2 == 0:  # False
# print("The number", number, "is odd")
# number = 4

# # iteration 3
# while number < 5:    # 4 < 5 => True
# if number % 2 == 0:  # True
# print("The number", number, "is even")
# number = 5

# # iteration 4
# while number < 5:    # 5 < 5 => False
# # while loop is done!
```

## Keywords: break and continue

### Python break statement

The 'break' statement terminates the loop containing it. Control of the program flows to the statement immediately after the body of the loop.

```python
# Use of break statement inside the loop
languages = ['C++', 'Python', 'Java']

for language in languages:
    if language == "Python":
        break
    print(language)

print("The end!")
```

### Python continue statement

The 'continue' statement is used to skip the rest of the code inside a loop for the current iteration only.

Loop does not terminate but continues on with the next iteration.

```python
# Use of continue statement inside the loop
languages = ['C++', 'Python', 'Java']

for language in languages:
    if language == "Python":
        continue
    print(language)

print("The end")
```

```python
# This program is same as the above example except the break statement has been replaced with continue.
```

### Function: range()

range() is a built-in Python function.

It is a very popular and widely used function in Python, especially when we are working with for loops and sometimes with while loops.

The range() function takes one or at most three arguments, namely the start and a stop value along with a step size.

The range() function can be represented in three different ways, or we can think of them as three range parameters:

> First way: range(stop)  # This by default considers the starting point as zero.

> Second way: range(start, stop)  # This generates the sequence based on the start and stop value.

> Third way: range(start, stop, step)  # It generates the sequence by incrementing the start value using the step size until it reaches the stop value.

```python
# Let's first check the type of the range() function.

# print(type(range(3)))  # type: range (not list)

print(list(range(3)))
print(list(range(10)))
print(list(range(1000 + 1)))
```
